```python
import pandas as pd
df = pd.read_csv("/content/spam_ham_dataset - spam_ham_dataset.csv")
df=df.drop("Unnamed: 0",axis=1)
print(df.head())
print("Shape:",df.shape)
print(df['label'].value_counts())
```

```python
# Features and Target
x_text=df['text']
y=df['label_num']
```

```python
# 2. Convert text -> numerical form (TF-IDF)

from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer(stop_words='english',
lowercase= True,
max_features= 5000,
ngram_range=(1,2))

x=vectorizer.fit_transform(x_text)
print("Feature Matrix Shape",df.shape)
```

```python
# 3. Split data (train/test)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
print("Train size:",x_train.shape[0],"|Test size:",x_test.shape[0])
```

```python
# 4. Apply Logistic Regression
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(
max_iter=1000,
C=1.0,
solver="liblinear")
model.fit(x_train,y_train)
```

```python
# 5. Evaluate accuracy
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
y_pred=model.predict(x_test)
acc=accuracy_score(y_test,y_pred)
print(f"\nAccuracy score: {acc:.4f}  ({acc*100:.2f}%)")
print("Accuracy:",acc)
print("\nClassification Report:\n",classification_report(y_test,y_pred))
print("\nConfusion Matrix:\n",confusion_matrix(y_test,y_pred))
```

```python
# Sample predictions
print("\n--- Sample predictions (first 10 test emails) ---")
label_map={0:"ham",1:"spam"}
for i in range(10):
  snippet=df.iloc[y_test.index[i]]['text'][:60].replace("\n"," ")
  actual=label_map[y_test.iloc[i]]
  pred=label_map[y_pred[i]]
  mark   = "✓" if actual == pred else "✗"
  print(f"{mark} Actual: {actual:4s} | Predicted: {pred:4s} | {snippet}...")
```
